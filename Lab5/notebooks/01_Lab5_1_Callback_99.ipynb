# Implementing Callbacks in TensorFlow using the MNIST Dataset

In the course you learned how to do classification using Fashion MNIST, a data set containing items of clothing. There's another, similar dataset called MNIST which has items of handwriting -- the digits 0 through 9.

Write an MNIST classifier that trains to 99% accuracy and stops once this threshold is achieved. In the past lectures you saw how this was done for the loss but here you will be using accuracy instead.

Some notes:
1. Your network should succeed in less than 9 epochs.
2. When it reaches 99% or greater it should print out the string "Reached 99% accuracy so cancelling training!" and stop training.
3. If you add any additional variables, make sure you use the same names as the ones used in the class. This is important for the function signatures (the parameters and names) of the callbacks.

```python
import os  # Підключаємо бібліотеку або інструмент: os.
import tensorflow as tf  # Підключаємо бібліотеку або інструмент: tensorflow as tf.
from tensorflow import keras  # Підключаємо бібліотеку або інструмент: tensorflow keras.
```

## Load and inspect the data

Begin by loading the data. A couple of things to notice:
 `load_data` returns the train and test sets in the form of the tuples `(x_train, y_train), (x_test, y_test)` but in this exercise you will be needing only the train set so you can ignore the second tuple.

```python
# Load the data


from keras.datasets import mnist  # Підключаємо бібліотеку або інструмент: keras.datasets mnist.


# Discard test set
(x_train, y_train), _ = tf.keras.datasets.mnist.load_data()  # Завантажуємо набір MNIST із зображеннями та правильними мітками.

# Normalize pixel values
x_train = x_train / 255.0  # Нормалізуємо яскравість пікселів до діапазону від нуля до одиниці.
```

Now take a look at the shape of the training data:

```python
data_shape = x_train.shape  # Обчислюємо та зберігаємо значення у змінній data_shape.

print(f"There are {data_shape[0]} examples with shape ({data_shape[1]}, {data_shape[2]})")  # Виводимо поточний результат або перевірочні значення.
```

## Defining your callback

Now it is time to create your own custom callback. For this complete the `myCallback` class and the `on_epoch_end` method in the cell below. If you need some guidance on how to proceed, check out this [link](https://www.tensorflow.org/guide/keras/custom_callback).

```python
# grader-required-cell

# GRADED CLASS: myCallback
### START CODE HERE
# Зупинка за навчальною точністю; тестова точність оцінюється окремо.
class myCallback(tf.keras.callbacks.Callback):  # Оголошуємо клас для власної логіки або зворотного виклику.
    def on_epoch_end(self, epoch, logs=None):  # Оголошуємо функцію; її назва показує, яку дію вона виконує.
        accuracy = (logs or {}).get("accuracy")  # Обчислюємо та зберігаємо значення у змінній accuracy.
        if accuracy is not None and accuracy >= 0.99:  # Перевіряємо умову й виконуємо вкладений код лише коли вона істинна.
            print("\nДосягнуто 99% точності, тому скасовуємо навчання!")  # Виводимо поточний результат або перевірочні значення.
            self.model.stop_training = True  # Обчислюємо та зберігаємо значення у змінній self.model.stop_training.
            os.makedirs("models", exist_ok=True)  # Обчислюємо та зберігаємо значення у змінній os.makedirs("models", exist_ok.
            self.model.save("models/task51_callback.keras")  # Зберігаємо навчену модель або результат на диск.
### END CODE HERE
```

## Create and train your model

Now that you have defined your callback it is time to complete the `train_mnist` function below.

**You must set your model to train for 10 epochs and the callback should fire before the 9th epoch for you to pass this assignment.**

**Hint:**
- Feel free to try the architecture for the neural network that you see fit but in case you need extra help you can check out an architecture that works pretty well at the end of this notebook.

```python
# grader-required-cell

# GRADED FUNCTION: train_mnist
def train_mnist(x_train, y_train):  # Оголошуємо функцію; її назва показує, яку дію вона виконує.

    ### START CODE HERE
    tf.keras.utils.set_random_seed(42)  # Фіксуємо генератор випадкових чисел для відтворюваного досліду.
    callbacks = myCallback()  # Обчислюємо та зберігаємо значення у змінній callbacks.
    model = tf.keras.models.Sequential([  # Обчислюємо та зберігаємо значення у змінній model.
        tf.keras.layers.Input(shape=(28, 28)),  # Задаємо форму вхідних даних для однієї цифри.
        tf.keras.layers.Flatten(),  # Перетворюємо карту ознак на один вектор перед повнозв’язними шарами.
        tf.keras.layers.Dense(512, activation="relu"),  # Додаємо повнозв’язний шар; останній шар формує оцінки десяти цифр.
        tf.keras.layers.Dense(10, activation="softmax")  # Додаємо повнозв’язний шар; останній шар формує оцінки десяти цифр.
    ])  # Завершуємо список параметрів або виклик, що почався вище.
    model.compile(optimizer="adam", loss="sparse_categorical_crossentropy",  # Обираємо функцію втрат для міток у форматі one-hot.
                  metrics=["accuracy"])  # Обчислюємо та зберігаємо значення у змінній metrics.
    history = model.fit(x_train, y_train, epochs=10, callbacks=[callbacks], verbose=2)  # Навчаємо модель на тренувальних даних і стежимо за перевірочною вибіркою.
    ### END CODE HERE

    return history  # Повертаємо результат функції до місця її виклику.
```

Call the `train_mnist` passing in the appropiate parameters to get the training history:

```python
# grader-required-cell

hist = train_mnist(x_train, y_train)  # Обчислюємо та зберігаємо значення у змінній hist.
```

If you see the message `Reached 99% accuracy so cancelling training!` printed out after less than 9 epochs it means your callback worked as expected.
